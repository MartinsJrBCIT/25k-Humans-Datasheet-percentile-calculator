```python

#separated installation, code below
import pandas
import time

name= input("Insert username here:")

print(f"Hello {name}!")
time.sleep(2)


data = pandas.read_csv('SOCR-HeightWeight.csv')
meanHeight = data['Height(Inches)'].mean()
medianHeight = data['Height(Inches)'].median()
modeHeight = data['Height(Inches)'].mode()[0]
meanWeight = data['Weight(Pounds)'].mean()
medianWeight = data['Weight(Pounds)'].median()
modeWeight = data['Weight(Pounds)'].mode()[0]

stdWeight = data['Weight(Pounds)'].std()
stdHeight = data['Height(Inches)'].std()

print(f"Mean (Height): {round(meanHeight, 2)}")
print(f"Median (Height): {round(medianHeight, 2)}")
print(f"Mode (Height): {round(modeHeight, 2)}")
print(f"Mean (Weight): {round(meanWeight,2)}")
print(f"Median (Weight): {round(medianWeight, 2)}")
print(f"Mode (Weight): {round(modeWeight, 2)}")
print(f"Standard Deviation (Weight): {round(stdWeight, 2)}")
print(f"Standard Deviation (Height): {round(stdHeight, 2)}")

heightData = sorted(data['Height(Inches)'])
weightData = sorted(data['Weight(Pounds)'])
time.sleep(2)
print(f"\n\n{name}, compare your self to others")

myHeight = float(input("Please input your height(Inches): "))

heightData.append(myHeight)
heightCalc = []

sortedHeightData = sorted(heightData)

for height in sortedHeightData:
    if height == myHeight:
        break
    else:
        heightCalc.append(height)

percent = (len(heightCalc)/len(heightData))*100
print(f"Your Perentile  (Height) is {round(percent, 2)}%")

time.sleep(2)

myWeight = float(input("Please input your weight(Pounds): "))

weightData.append(myWeight)
weightCalc = []

sortedWeightData = sorted(weightData)

for weight in sortedWeightData:
    if weight == myWeight:
        break
    else:
        weightCalc.append(weight)

percent = (len(weightCalc)/len(weightData))*100
print(f"{name},our Perentile (Weight) is {round(percent, 2)}%")
```
